#  Apa itu Linear Model?

Model Linear mencoba menarik satu garis lurus yang paling pas mewakili semua titik tersebut. Secara matematis, garis prediksi ($\hat{y}$) dirumuskan sebagai kombinasi penjumlahan dari semua fitur ($x$) yang dikalikan dengan bobotnya ($w$), kemudian Linear model mengasumsikan:

$\hat{y} = w_1x_1 + w_2x_2 + ... + w_nx_n + b$

Artinya:

* x = fitur (input)
* w = bobot (coef_)
* b = bias (intercept_)
* ŷ = hasil prediksi

Di scikit-learn:

* `coef_` → bobot
* `intercept_` → bias

Linear model bisa dipakai untuk:

* Regresi (prediksi angka)
* Klasifikasi (prediksi kelas)
