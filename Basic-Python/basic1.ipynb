## String Formatting di Python

String formatting adalah cara untuk **menyisipkan nilai (variabel) ke dalam sebuah string**.  
Python menyediakan beberapa metode string formatting, di antaranya menggunakan **operator `%`** dan **f-string**.

### 1. String Formatting dengan Operator `%`

```python
nama = "Andi"
umur = 20

print("Nama saya %s dan umur saya %d tahun" % (nama, umur))
````

Output:

```
Nama saya Andi dan umur saya 20 tahun
```

Keterangan:

* `%s` → untuk string
* `%d` → untuk bilangan bulat (integer)

---

### 2. String Formatting dengan Float

```python
waktu = 2.34567
print("Waktu eksekusi: %.2f detik" % waktu)
```

Output:

```
Waktu eksekusi: 2.35 detik
```

Keterangan:

* `%.2f` → menampilkan angka desimal dengan 2 digit

---

### 3. String Formatting dengan f-string (Direkomendasikan)

```python
nama = "Budi"
umur = 22

print(f"Nama saya {nama} dan umur saya {umur} tahun")
```

Output:

```
Nama saya Budi dan umur saya 22 tahun
```

### Kesimpulan

* String formatting digunakan untuk menggabungkan teks dengan variabel
* Operator `%` adalah cara lama namun masih sering digunakan
* **f-string lebih sederhana, mudah dibaca, dan direkomendasikan di Python modern**

```



`````python
# Formating dengan operator "%"
name = "Sharlen"
age  = 22
jk = "Perempuan"
print("Nama saya %s, umur saya %d tahun, saya berjenis kelamin %s" % (name, age, jk))
# Formating dengan method .format()
print("Nama saya {0}, umur saya {1} tahun, saya berjenis kelamin {2}".format(name, age, jk))
`````

`````python
# String Formatting dengan float
num = 1/3 # hasilnya 0.3333333 dst
print("Nilai num adalah %.2f" % num)
print("Nilai num adalah {:.2f}".format(num))
`````

`````python
# Formating dengan f-string (Direkomendasikan di Python 3.6 ke atas)
name = "Sharlen"
age  = 22
jk = "Perempuan"
print(f"Nama saya {name}, umur saya {age} tahun, saya berjenis kelamin {jk}")
`````

## Assignment Nilai dan Penamaan Variabel

Pada bagian ini, kita akan mempelajari **cara memberikan nilai ke variabel (assignment)**,  
**mengubah nilai variabel**, serta **aturan penamaan variabel** dalam Python.

Assignment adalah proses **menyimpan suatu nilai ke dalam variabel** menggunakan tanda `=`.

---

### 1. Assignment Nilai

Assignment nilai digunakan untuk **memberikan nilai awal ke sebuah variabel**.

```python
a = 10
x = 5
````

Pada contoh di atas:

* Variabel `a` menyimpan nilai `10`
* Variabel `x` menyimpan nilai `5`

---

### 2. Pemanggilan Variabel

Pemanggilan variabel dilakukan untuk **menampilkan nilai yang tersimpan di dalam variabel** menggunakan fungsi `print()`.

```python
print("Nilai a = ", a)
print("Nilai x = ", x)
```

Output:

```
Nilai a =  10
Nilai x =  5
```

---

### 3. Assignment dengan Operator (`+=`)

Assignment dengan operator digunakan untuk **mengubah nilai variabel berdasarkan nilai sebelumnya**.

```python
x += a  # sama dengan x = x + a
```

Artinya:

* Nilai `x` ditambah dengan nilai `a`
* Hasilnya disimpan kembali ke variabel `x`

```python
print("Nilai x = ", x)
```

Output:

```
Nilai x =  15
```

---

### 4. Penamaan Variabel

Penamaan variabel harus mengikuti aturan Python dan sebaiknya menggunakan nama yang jelas dan mudah dipahami.

Contoh penamaan variabel yang benar:

```python
nilai_y = 20      # penamaan dengan underscore (snake_case)
nilaiZ = 30       # penamaan dengan camelCase
juta1 = 1000000   # penamaan dengan angka di akhir
```

Aturan umum penamaan variabel:

* Tidak boleh diawali angka
* Tidak boleh menggunakan spasi
* Boleh menggunakan huruf, angka, dan underscore (`_`)

---

### 5. Assignment Tidak Langsung (Indirect Assignment)

Assignment tidak langsung adalah proses **memberikan nilai dari satu variabel ke variabel lain**.

```python
b = a
print("Nilai b = ", b)
```

Output:

```
Nilai b =  10
```

Artinya:

* Variabel `b` mengambil nilai yang sama dengan variabel `a`

`````python
# Menaru/asigment nilai 
a  = 10
x  = 5
`````

`````python

# pemanggilan pertama
print("Nilai a = ", a)
print("Nilai x = ", x)
x += a  # x = x + a
# pemanggilan kedua
print("Nilai a = ", x)
`````

`````python
# penamaan 
nilai_y = 20 # contoh penamaan dengan underscore
nilaiZ = 30  # contoh penamaan dengan camel case
juta1 = 1000000 # contoh penamaan dengan angka di akhir
`````

`````python
# assigment indirect
b = a
print("Nilai b = ", b)
`````
